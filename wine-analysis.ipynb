```python
plt.boxplot(df["alcohol"])
plt.title("Box Plot of Alcohol")
plt.ylabel("Alcohol")
plt.show()
```

```python
plt.boxplot(df["proline"])
plt.title("Box Plot of proline")
plt.ylabel("Proline")
plt.show()
```

The color_intensity feature has more visible possible outliers than the other features.

The proline feature has long whiskers indicating that its non-outlier values cover a wide range.

```python
#3
Q1 = df["alcohol"].quantile(0.25)
Q3 = df["alcohol"].quantile(0.75)

print("Q1:", Q1)
print("Q3:", Q3)
```

```python
#4
IQR = Q3 - Q1
print("IQR:", IQR)
```

```python
#5
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
#6
outliers = df[(df["alcohol"] < lower_bound) | (df["alcohol"] > upper_bound)]
print("Number of outliers:", len(outliers))
```

#7 
not all outliers should be error and removed it should be investigated before taking any action . if it is comfirmed to be error it should be corrected or removed
